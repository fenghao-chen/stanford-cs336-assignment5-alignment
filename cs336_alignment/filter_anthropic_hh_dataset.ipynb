```python
def filter(path: str) -> list[str]:
    data = []
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            if line.count("Human:") == 2:
                data.append(line)

    return data

result = []
result += filter("Anthropic_hh-rlhf/harmless-base/train.jsonl")
result += filter("Anthropic_hh-rlhf/helpful-base/train.jsonl")
result += filter("Anthropic_hh-rlhf/helpful-online/train.jsonl")
result += filter("Anthropic_hh-rlhf/helpful-rejection-sampled/train.jsonl")
```

```python
import random
random.shuffle(result)
```

```python
with open("anthropic_hh.jsonl", "w") as file:
    file.writelines(result)
```
